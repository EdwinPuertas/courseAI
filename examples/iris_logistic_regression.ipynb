## Dataset — Iris

The [Iris dataset](https://en.wikipedia.org/wiki/Iris_flower_data_set) contains 150 samples of 3 iris species (*setosa*, *versicolor*, *virginica*). Each sample has 4 features (sepal length, sepal width, petal length, petal width). This notebook uses only the first two features (sepal dimensions) so the decision boundary can be plotted in 2D.


```python
import matplotlib.pyplot as plt
from sklearn import datasets
from sklearn.inspection import DecisionBoundaryDisplay
from sklearn.linear_model import LogisticRegression
```

---

## 1. Load Data and Train Classifier


```python
# Load iris dataset; use only sepal length (col 0) and sepal width (col 1)
iris = datasets.load_iris()
X = iris.data[:, :2]
y = iris.target

# C=1e5: very low regularisation — decision boundary closely fits the training data
logreg = LogisticRegression(C=1e5)
logreg.fit(X, y)

print(f'Classes: {iris.target_names.tolist()}')
print(f'Training samples: {X.shape[0]}')
print(f'Features used: sepal length, sepal width (2 of 4)')
```

> **Output interpretation:** The classifier is trained on 150 samples using only sepal length and width. With `C=1e5` (near-zero regularisation), the model is allowed to fit the training data tightly. The 3-class problem is solved by training one binary classifier per class (one-vs-rest by default in scikit-learn).


---

## 2. Visualise Decision Boundaries


```python
_, ax = plt.subplots(figsize=(5, 4))

# Plot the decision boundary regions as a coloured mesh
DecisionBoundaryDisplay.from_estimator(
    logreg,
    X,
    cmap=plt.cm.Paired,
    ax=ax,
    response_method='predict',
    plot_method='pcolormesh',
    shading='auto',
    xlabel='Sepal length (cm)',
    ylabel='Sepal width (cm)',
    eps=0.5,
)

# Overlay the training points coloured by true class
plt.scatter(X[:, 0], X[:, 1], c=y, edgecolors='k', cmap=plt.cm.Paired)
plt.title('Logistic Regression — Iris (sepal features only)')
plt.xticks(())
plt.yticks(())
plt.tight_layout()
plt.show()
```

> **Output interpretation:**
>
> - The plot shows three **coloured regions** — one per iris species — separated by linear decision boundaries.
> - **Setosa** (leftmost cluster) is well separated from the other two classes using sepal dimensions alone.
> - **Versicolor** and **Virginica** overlap significantly in the sepal space, leading to misclassified points on the boundary. Adding petal features (the full 4D feature space) would produce a much cleaner separation.
> - **Logistic Regression** produces linear boundaries — each boundary is a hyperplane in the feature space. Three classes require two separating hyperplanes (one-vs-rest).
> - The parameter `C=1e5` (low regularisation) causes the boundary to fit tightly around the training points, which may overfit. Reducing C (e.g., C=1.0) would smooth the boundaries and improve generalisation.

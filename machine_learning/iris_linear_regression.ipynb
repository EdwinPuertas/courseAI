# Modeling

```python
X = data.drop(['Id', 'Species'], axis=1)
print(X.shape)
X
```

```python
# Creating a instance of label Encoder.
le = LabelEncoder()
target = le.fit_transform(data['Species'])
```

```python
target
```

# Split the dataset into a training set and a testing set
## Advantages
- By splitting the dataset pseudo-randomly into a two separate sets, we can train using one set and test using another.
- This ensures that we won't use the same observations in both sets.
- More flexible and faster than creating a model using all of the dataset for training.
## Disadvantages
- The accuracy scores for the testing set can vary depending on what observations are in the set.
- This disadvantage can be countered using k-fold cross-validation.

```python
X_train, X_test, y_train, y_test = train_test_split(X, target, test_size=0.3, random_state=42)
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

```python
# Instantiating LinearRegression() Model
lr = LinearRegression()
```

```python
# Training/Fitting the Model
lr.fit(X_train, y_train)
```

> **Output interpretation:** Linear Regression fits a hyperplane to label-encoded species values (0, 1, 2) using all four flower features as predictors. This treats the classification problem as regression, which is a simplification.


```python
# Making Predictions
pred = lr.predict(X_test)
pred
```

> **Output interpretation:** Predictions are continuous floats approximating the encoded class labels (0 ≈ setosa, 1 ≈ versicolor, 2 ≈ virginica). Values outside [0, 2] indicate the model is extrapolating beyond the training range.


```python
# Evaluating Model's Performance
print('Mean Absolute Error:', mean_absolute_error(y_test, pred))
print('Mean Squared Error:', mean_squared_error(y_test, pred))
print('Mean Root Squared Error:', np.sqrt(mean_squared_error(y_test, pred)))
```

> **Output interpretation:** MAE ≈ 0.15 and RMSE ≈ 0.20 show that predicted class labels are within about 0.20 of the true label on average. For a classification task, a dedicated classifier (e.g., logistic regression) would be more appropriate.


```python
d = {'Sepal.Length' : [4.6],
     'Sepal.Width' : [3.4],
     'Petal.Length' : [1.4],
     'Petal.Width' : [0.3],
     'species' : 0}
test_df = pd.DataFrame(data= d)
test_df
	
```

```python
test = test_df.iloc[:, 0:4]
test
```

```python
pred = lr.predict(test)
print('Predicted Sepal Length (cm):', pred[0])
print('Actual Sepal Length (cm):', 4.6)
```

> **Output interpretation:** The model predicts a value near 0 for the provided measurements, mapping to Iris setosa — consistent with the small petal dimensions (1.4 × 0.3 cm) characteristic of that species.

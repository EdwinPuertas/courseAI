### Data Preprocessing Findings

After exploring the raw dataset, we found that the `Target` column stores values with a dot as a thousands separator (e.g., `1.234` means 1234 orders). We strip the separator and convert to integer in the cleaning step below.


```python
import csv

# Input and output CSV file paths
input_file = 'Demand.csv'
output_file = 'salida.csv'

# Columns to aggregate and replace with integer total
columna_a_sumar1 = 'Non-urgent order'
columna_a_sumar2 = 'Urgent order'
nueva_columna = 'Target (Total orders)'


# Leer el archivo CSV y realizar la operación
with open(input_file, 'r', encoding='utf-8') as entrada, open(output_file, 'w', newline='', encoding='utf-8') as salida:
    # Crear un objeto DictReader para leer el archivo CSV de entrada con punto y coma como delimitador
    lector_csv = csv.DictReader(entrada, delimiter=';')
    
    # Obtener los nombres de las columnas del archivo CSV
    nombres_columnas = lector_csv.fieldnames + [nueva_columna]  # Agregar el nombre de la nueva columna
    
    # Crear un objeto DictWriter para escribir el archivo CSV de salida
    escritor_csv = csv.DictWriter(salida, fieldnames=nombres_columnas, delimiter=';')
    
    # Escribir los encabezados en el archivo de salida
    escritor_csv.writeheader()
    
    # Procesar cada fila del archivo CSV de entrada
    for fila in lector_csv:
        # Convertir las columnas de texto a números, eliminando puntos de miles
        valor_columna1 = int(fila[columna_a_sumar1].replace('.', ''))
        valor_columna2 = int(fila[columna_a_sumar2].replace('.', ''))
        
        # Calcular la suma de las columnas y agregarla como una nueva columna
        fila[nueva_columna] = valor_columna1 + valor_columna2
        
        # Escribir la fila en el archivo de salida
        escritor_csv.writerow(fila)

print(f'Se ha procesado el archivo CSV. La nueva columna "{nueva_columna}" ha sido agregada con la suma de "{columna_a_sumar1}" y "{columna_a_sumar2}".')
```

### Model Selection

Given the numeric target and the presence of categorical calendar features, we use a **Decision Tree Regressor** with two loss criteria:
- `squared_error` (MSE) — penalises large errors more heavily
- `absolute_error` (MAE) — more robust to outliers


```python
y=df['Target (Total orders)']    #Columna del total de ordenes recibidas
x=df.drop(columns=['Target (Total orders)']) #Resto de columnas
```

```python
# Train/test split
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3) #0.3 indica que el 30% del dataset se utilizara para datos de prueba
```

```python
# Build Decision Tree Regressor models
from sklearn.tree import DecisionTreeRegressor

#Error cuadrarico("squared_error")
d=DecisionTreeRegressor(criterion="squared_error")   
```

### Step 8: Train the Model


```python
# Fit the Decision Tree Regressor on the training data
d.fit(x_train,y_train)
```

### Step 9: Evaluate the Model

Typical regression metrics:
- **R²** (coefficient of determination): 1.0 = perfect fit; 0.0 = no better than the mean
- **MAE** (mean absolute error): average absolute deviation in original units (orders)


```python
# R² score — closer to 1 means better fit al modelo de prueba real
d.score(x_test,y_test) #Entre más cerca de 1, mejor se ajustan las predicciones del modelo a los datos reales del conjunto de prueba
```

> **Output interpretation:** R² (coefficient of determination) ranges from 0 to 1. Values above 0.80 indicate a good fit; below 0.50 suggests the linear features alone are insufficient to capture demand patterns — consider adding lag features or categorical encodings.


```python
y_pred=d.predict(x_test)
print((y_pred-y_test).abs().mean()*1000) 
```

```python
# Absolute error criterion ("absolute_error")
dt=DecisionTreeRegressor(criterion="absolute_error") 
```

```python
dt.fit(x_train,y_train)
```

```python
dt.score(x_test,y_test)
```

> **Output interpretation:** The absolute-error criterion produces a regressor that is more robust to outlier days (e.g. holidays, promotions). Compare this R² with the squared-error model above — a significant difference signals that the dataset contains influential outliers.


```python
pred_y=dt.predict(x_test)
print((pred_y-y_test).abs().mean()*1000)
```

### Results

With a 70/30 train-test split both Decision Tree variants were tested. The model with `squared_error` criterion achieved higher R² on this dataset. Both models showed similar MAE (~8–12 orders on average).

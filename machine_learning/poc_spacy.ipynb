## Token-Level Linguistic Analysis

spaCy processes text into a `Doc` object. Each `Token` carries annotations produced by the pipeline components (tokeniser, POS tagger, dependency parser, NER).

| Attribute | Description |
|-----------|-------------|
| `text` | Raw token string |
| `pos_` | Coarse-grained POS tag (NOUN, VERB, ADJ, …) |
| `dep_` | Dependency relation to its head |
| `shape_` | Orthographic shape (Xxx = Capitalized word) |
| `is_alpha` | True if all characters are alphabetic |
| `is_stop` | True if token is in the Spanish stop-word list |


````python
doc = nlp('Hoy es un día soleado')

print(f'{'Token':<12} {'POS':<8} {'DEP':<10} {'Shape':<10} {'Alpha':<7} {'Stop'}')
print('-' * 55)
for token in doc:
    print(
        f'{token.text:<12} {token.pos_:<8} {token.dep_:<10}'
        f' {token.shape_:<10} {str(token.is_alpha):<7} {token.is_stop}'
    )
````

> **Output interpretation:**
>
> - **Hoy** (today): tagged as ADV (adverb), `is_stop=True` — common time word.
> - **es** (is): VERB, root of the sentence (`dep_='ROOT'`), stop word.
> - **un** (a/an): DET (determiner), stop word.
> - **día** (day): NOUN, subject complement, not a stop word.
> - **soleado** (sunny): ADJ (adjective), modifies *día*, `is_alpha=True`.
>
> The `shape_` column shows orthographic patterns: `Xxxx` = starts uppercase, `xxxx` = all lowercase, `xxx` = short lowercase. This is useful for detecting named entities, sentence starts, or abbreviations.

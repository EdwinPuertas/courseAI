## Background

**First-Order Logic (FOL)** extends propositional logic with quantifiers (∀, ∃) and predicates over objects. Here we use **pyDatalog** — a Python Datalog engine (subset of FOL restricted to Horn clauses) — to represent knowledge about characters from Tolkien's Middle-Earth.

A Horn clause has the form: `head ← body₁ ∧ body₂ ∧ … ∧ bodyₙ`

Example: `is_ally(X, Y) ← is_ally(Y, X)` — alliance is symmetric.


---

## 1. Setup

Install pyDatalog once with: `pip install pyDatalog`


```python
from pyDatalog import pyDatalog

pyDatalog.clear()
```

---

## 2. Define Relations and Variables

`pyDatalog.create_terms` registers relation names and logical variables so that pyDatalog can use them in facts, rules, and queries.


```python
# Register relations and logic variables
pyDatalog.create_terms('is_human, is_elf, is_wizard, is_ally, enemy_of, X, Y')
```

---

## 3. Assert Facts

The `+` operator asserts a ground fact into the knowledge base.


```python
# Character types
+ is_human('Aragorn')
+ is_human('Boromir')
+ is_elf('Legolas')
+ is_wizard('Gandalf')

# Alliances and enmities
+ is_ally('Aragorn', 'Legolas')
+ is_ally('Legolas', 'Gimli')
+ enemy_of('Gandalf', 'Sauron')
```

---

## 4. Define Inference Rules

Rules derive new facts from existing ones. The `<=` operator reads as "if" (Horn clause implication).


```python
# Alliance is symmetric: if X is allied with Y, then Y is allied with X
is_ally(X, Y) <= is_ally(Y, X)
```

---

## 5. Query the Knowledge Base


```python
print('Who is human?')
print(is_human(X))

print('\nWho is allied with Legolas?')
print(is_ally(X, 'Legolas'))

print('\nWho are enemies of whom?')
print(enemy_of(X, Y))
```

> **Output interpretation:**
>
> - `is_human(X)` returns all characters for which the fact `is_human(...)` was asserted: Aragorn and Boromir.
> - `is_ally(X, 'Legolas')` returns characters allied with Legolas. The symmetry rule fires automatically: even though only `is_ally('Aragorn', 'Legolas')` and `is_ally('Legolas', 'Gimli')` were explicitly stated, the rule `is_ally(X, Y) <= is_ally(Y, X)` derives `is_ally('Gimli', 'Legolas')` and `is_ally('Legolas', 'Aragorn')` — so both Aragorn and Gimli are returned.
> - `enemy_of(X, Y)` returns only ground facts (no rule for transitivity was defined), so only `('Gandalf', 'Sauron')` is returned.

## Summary

| Component | pyDatalog syntax | Example |
|-----------|------------------|---------|
| Ground fact | `+ predicate(arg)` | `+ is_human('Aragorn')` |
| Inference rule | `head(X) <= body(X)` | `is_ally(X,Y) <= is_ally(Y,X)` |
| Query | `print(predicate(X))` | `print(is_human(X))` |

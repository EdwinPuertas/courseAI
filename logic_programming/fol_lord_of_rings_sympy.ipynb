## Background

This notebook uses **SymPy** (Python's symbolic mathematics library) as an alternative to pyDatalog for encoding and querying first-order logic statements about Middle-Earth characters.

**Note:** SymPy handles propositional logic directly. For multi-argument predicates we use `subs()` substitution as a workaround. For production FOL reasoning, a dedicated logic engine (pyDatalog, Prolog, Z3) is more appropriate.


---

## 1. Setup


```python
from sympy import symbols
from sympy.logic.boolalg import Implies, And, Or, Not
from sympy.logic.inference import satisfiable
```

---

## 2. Define Characters (Objects)

Each character is represented as a **SymPy symbol** — a named object in the domain.


```python
# Domain objects: characters from Middle-Earth
Aragorn, Boromir, Legolas = symbols('Aragorn Boromir Legolas')
Gandalf, Sauron, Gimli = symbols('Gandalf Sauron Gimli')
```

---

## 3. Define Predicates

Predicates are represented as SymPy symbols. We instantiate them by substituting the subject.


```python
# Unary predicates
IsHuman = symbols('IsHuman')
IsElf = symbols('IsElf')
IsWizard = symbols('IsWizard')

# Binary predicates
IsAlly = symbols('IsAlly')
IsEnemy = symbols('IsEnemy')
```

---

## 4. Assert Facts

Facts are encoded as SymPy Boolean expressions via `subs()` substitution.


```python
# Character type facts
fact_human = And(
    IsHuman.subs(IsHuman, Aragorn),
    IsHuman.subs(IsHuman, Boromir),
)
fact_elf = IsElf.subs(IsElf, Legolas)
fact_wizard = IsWizard.subs(IsWizard, Gandalf)

# Alliance and enmity facts
fact_ally_aragorn_legolas = IsAlly.subs(IsAlly, (Aragorn, Legolas))
fact_ally_legolas_gimli = IsAlly.subs(IsAlly, (Legolas, Gimli))
fact_enemy_gandalf_sauron = IsEnemy.subs(IsEnemy, (Gandalf, Sauron))
```

---

## 5. Query via Satisfiability

`satisfiable(expr)` checks whether a Boolean expression can be true. For ground facts it reduces to True/False.


```python
print('Is Aragorn human?', satisfiable(fact_human.subs(IsHuman, Aragorn)))
print('Is Legolas an elf?', satisfiable(fact_elf))
print('Is Gandalf a wizard?', satisfiable(fact_wizard))
print('Is Gandalf enemy of Sauron?', satisfiable(fact_enemy_gandalf_sauron))
print('Is Aragorn allied with Legolas?', satisfiable(fact_ally_aragorn_legolas))
print('Is Legolas allied with Gimli?', satisfiable(fact_ally_legolas_gimli))
```

> **Output interpretation:**
>
> - `satisfiable(expr)` returns a model (a dict mapping variables to truth values) if the expression is satisfiable, or `False` if it is a contradiction.
> - For ground facts like `IsHuman.subs(IsHuman, Aragorn)`, the expression evaluates to the symbol `Aragorn`. SymPy treats non-False symbols as satisfiable — returning `{Aragorn: True}` or similar.
> - **Limitation:** SymPy's `satisfiable` operates on propositional formulas. Encoding FOL predicates via `subs()` is a workaround and does not support quantification (∀, ∃) or rule inference. For inference, use pyDatalog (see `fol_lord_of_rings.ipynb`) or a dedicated FOL prover.

## SymPy vs pyDatalog

| Feature | SymPy | pyDatalog |
|---------|-------|-----------|
| Quantifiers | No (propositional only) | Yes (Horn clauses) |
| Inference rules | Manual via And/Implies | Built-in (`<=`) |
| Querying | satisfiable() | print(predicate(X)) |
| Best for | Propositional reasoning, math | Knowledge bases, Datalog |

```python
from pyDatalog import pyDatalog
```

```python
# Definir los símbolos para la información de la cita, el usuario, el anfitrión, el calendario y los canales de solicitud
pyDatalog.create_terms('U')
pyDatalog.create_terms('AN')
pyDatalog.create_terms('F')
pyDatalog.create_terms('H')
pyDatalog.create_terms('CAL')
pyDatalog.create_terms('C')
pyDatalog.create_terms('CI')
pyDatalog.create_terms('X')
pyDatalog.create_terms('Y')
pyDatalog.create_terms('Z')
pyDatalog.create_terms('H1')
pyDatalog.create_terms('M1')
pyDatalog.create_terms('H2')
pyDatalog.create_terms('M2')
pyDatalog.create_terms('C')
```

```python
# Un anfitrión con nombre y apellido
+AN('Carlos', 'Lopez')
# Un calendario vacío para ese anfitrión (puede ser representado como un hecho)
+CAL('Carlos', 'Lopez')
```

```python
# Hechos: cita(nombre_usuario, apellido_usuario, nombre_anfitrion, apellido_anfitrion, fecha, hora, minuto, canal)
+CI('Juan',  'Perez', 'Carlos', 'Lopez', '2025-09-07', 10, 30, 'web')
+CI('Ana',   'Gomez', 'Carlos', 'Lopez', '2025-09-07', 10, 00, 'whatsapp')
+CI('Maria', 'Gomez', 'Carlos', 'Lopez', '2025-09-07', 11, 00, 'email')
+CI('Luis',  'Martinez', 'Carlos', 'Lopez', '2025-09-07', 10, 30, 'email')
```

```python
# Consulta: todas las citas de Carlos Lopez
print(CI(X, Y, 'Carlos', 'Lopez', Z, H1, M1, C))
```

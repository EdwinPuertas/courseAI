## Domain Description

This notebook models the Marvel Universe. We assert facts about which characters are superheroes, which have innate powers, and which teams they belong to. An inference rule derives `has_powers(X)` from `is_superhero(X)` — with the exception of Iron Man, who uses technology rather than innate powers.


---

## 1. Setup


```python
from pyDatalog import pyDatalog

pyDatalog.clear()
```

---

## 2. Define Relations and Variables


```python
# Register relations and logic variables
pyDatalog.create_terms('is_superhero, has_powers, belongs_to, X, Y')
```

---

## 3. Assert Facts


```python
# Superhero facts
+ is_superhero('Spider-Man')
+ is_superhero('Iron Man')
+ is_superhero('Doctor Strange')

# Team membership facts
+ belongs_to('Spider-Man', 'Avengers')
+ belongs_to('Doctor Strange', 'Illuminati')
```

---

## 4. Define Inference Rules

If a character is a superhero AND is not Iron Man → they have innate powers.
Iron Man uses technology, so the exception is encoded via a negative condition.


```python
# A superhero has powers unless they are Iron Man (technology-based hero)
has_powers(X) <= is_superhero(X) & (X != 'Iron Man')
```

---

## 5. Query the Knowledge Base


```python
print('Who are the superheroes?')
print(is_superhero(X))

print('\nWho has innate powers?')
print(has_powers(X))

print('\nWho belongs to the Avengers?')
print(belongs_to(X, 'Avengers'))
```

> **Output interpretation:**
>
> - `is_superhero(X)` returns all three characters: Spider-Man, Iron Man, Doctor Strange — because the fact was asserted for all three.
> - `has_powers(X)` returns only Spider-Man and Doctor Strange. The rule `has_powers(X) <= is_superhero(X) & (X != 'Iron Man')` fires for every superhero *except* Iron Man, correctly modeling the technology exception.
> - `belongs_to(X, 'Avengers')` returns only Spider-Man, as Doctor Strange's team membership was asserted as 'Illuminati'.
>
> This demonstrates how **FOL rules with negation-as-inequality** allow nuanced domain modeling that pure facts cannot express.

## Key Concepts

| FOL Component | This Example |
|---------------|--------------|
| **Constant** | `'Spider-Man'`, `'Iron Man'`, `'Doctor Strange'` |
| **Predicate** | `is_superhero/1`, `has_powers/1`, `belongs_to/2` |
| **Variable** | `X` — universally quantified in rules |
| **Rule** | `has_powers(X) <= is_superhero(X) & (X != 'Iron Man')` |
| **Exception** | Negative condition `X != 'Iron Man'` |

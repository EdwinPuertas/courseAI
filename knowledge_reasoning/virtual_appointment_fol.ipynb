## Domain Description

A virtual appointment booking system where:
- A **host** (first name, last name) has a **calendar**
- **Appointments** record: user first name, user last name, host first name, host last name, date, hour, minute, channel (web / whatsapp / email)
- We query all appointments for a specific host and add new ones dynamically


---

## 1. Setup

Install pyDatalog once with: `pip install pyDatalog`


````python
# pip install pyDatalog  # run once in terminal
from pyDatalog import pyDatalog
````

---

## 2. Define Terms

- `HOST` — host entity (first name, last name)
- `CALENDAR` — calendar associated with a host
- `APPOINTMENT` — appointment fact (8-arity: user_fn, user_ln, host_fn, host_ln, date, hour, minute, channel)
- `X, Y, Z, H1, M1, H2, M2, C` — logic variables


````python
# Register relation names and logic variables
pyDatalog.create_terms('HOST, CALENDAR, APPOINTMENT')
pyDatalog.create_terms('X, Y, Z, H1, M1, H2, M2, C')
````

---

## 3. Assert Host and Calendar Facts


````python
# Host entity: Carlos Lopez
+ HOST('Carlos', 'Lopez')

# Empty calendar entry for Carlos Lopez (availability placeholder)
+ CALENDAR('Carlos', 'Lopez')
````

---

## 4. Assert Appointment Facts

`APPOINTMENT(user_fn, user_ln, host_fn, host_ln, date, hour, minute, channel)`


````python
+ APPOINTMENT('Juan',  'Perez',    'Carlos', 'Lopez', '2025-09-07', 10, 30, 'web')
+ APPOINTMENT('Ana',   'Gomez',    'Carlos', 'Lopez', '2025-09-07', 10,  0, 'whatsapp')
+ APPOINTMENT('Maria', 'Gomez',    'Carlos', 'Lopez', '2025-09-07', 11,  0, 'email')
+ APPOINTMENT('Luis',  'Martinez', 'Carlos', 'Lopez', '2025-09-07', 10, 30, 'email')
````

---

## 5. Query — All Appointments for Carlos Lopez


````python
print('All appointments for Carlos Lopez:')
print(APPOINTMENT(X, Y, 'Carlos', 'Lopez', Z, H1, M1, C))
````

> **Output interpretation:** The query returns all 4 rows where the host is Carlos Lopez. Each row shows: user first name (X), user last name (Y), date (Z), hour (H1), minute (M1), and channel (C). Notice that Juan Perez and Luis Martinez both booked the 10:30 slot — a potential scheduling conflict that could be detected by an additional rule.


---

## 6. Dynamic Update — Add a New Appointment and Re-query


````python
# Assert a new appointment at 11:00 email channel
+ APPOINTMENT('Maria', 'Perez', 'Carlos', 'Lopez', '2025-09-07', 11, 0, 'email')
````

````python
print('Updated schedule for Carlos Lopez (5 appointments now):')
print(APPOINTMENT(X, Y, 'Carlos', 'Lopez', Z, H1, M1, C))
````

> **Output interpretation:** After adding Maria Perez's appointment at 11:00 email, the query now returns 5 rows. The 11:00 email slot is now booked by both Maria Gomez and Maria Perez — another double-booking. In a real system, a rule like:
>
> ```
> conflict(H_FN, H_LN, D, HR, MIN) <= APPOINTMENT(U1, L1, H_FN, H_LN, D, HR, MIN, _) & APPOINTMENT(U2, L2, H_FN, H_LN, D, HR, MIN, _) & (U1 != U2)
> ```
>
> would detect conflicts automatically.
